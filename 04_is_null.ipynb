### 1. 경기도에 위치한 식품창고 목록 출력하기

```python
SELECT WAREHOUSE_ID, WAREHOUSE_NAME, ADDRESS, IFNULL(FREEZER_YN, "N")
FROM FOOD_WAREHOUSE
WHERE ADDRESS LIKE '경기도%'
ORDER BY WAREHOUSE_ID ASC
```

### 2. 이름이 없는 동물의 아이디

```python
SELECT ANIMAL_ID
FROM ANIMAL_INS
WHERE NAME IS NULL
ORDER BY ANIMAL_ID ASC
```

### 3. 이름이 있는 동물의 아이디

```python
SELECT ANIMAL_ID
FROM ANIMAL_INS
WHERE NAME IS NOT NULL
ORDER BY ANIMAL_ID ASC
```

### 4. NULL 처리하기

```python
SELECT ANIMAL_TYPE, IFNULL(NAME, "No name"), SEX_UPON_INTAKE
FROM ANIMAL_INS
```

### 5. 나이 정보가 없는 회원 수 구하기

```python
SELECT COUNT(USER_ID) AS USERS
FROM USER_INFO
WHERE AGE IS NULL
```

### 6. ROOT 아이템 구하기

```python
SELECT T.ITEM_ID, I.ITEM_NAME
FROM ITEM_TREE AS T
JOIN ITEM_INFO AS I ON T.ITEM_ID = I.ITEM_ID
WHERE T.PARENT_ITEM_ID IS NULL
ORDER BY T.ITEM_ID
```

### **<span style="color:#FFAEC9">7. 업그레이드 할 수 없는 아이템 구하기</span>**

```python
SELECT I.ITEM_ID, I.ITEM_NAME, I.RARITY
FROM ITEM_INFO AS I
JOIN ITEM_TREE AS T ON I.ITEM_ID = T.ITEM_ID
WHERE T.ITEM_ID NOT IN (
    SELECT PARENT_ITEM_ID
    FROM ITEM_TREE
    WHERE PARENT_ITEM_ID IS NOT NULL
)
ORDER BY I.ITEM_ID DESC
```

```python
SELECT I.ITEM_ID, I.ITEM_NAME, I.RARITY
FROM ITEM_INFO AS I
JOIN ITEM_TREE AS T ON I.ITEM_ID = T.ITEM_ID
WHERE NOT EXISTS (
    SELECT 1
    FROM ITEM_TREE AS T2
    WHERE T2.PARENT_ITEM_ID = T.ITEM_ID
)
ORDER BY I.ITEM_ID DESC
```

### 8. 잡은 물고기의 평균 길이 구하기

```python
SELECT ROUND(AVG(IFNULL(LENGTH, 10)), 2)
FROM FISH_INFO
```
