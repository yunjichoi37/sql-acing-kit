### **<span style="color:#FFAEC9">1. 주문량이 많은 아이스크림들 조회하기</span>**

```python
SELECT F.FLAVOR
FROM FIRST_HALF AS F
JOIN (
    SELECT FLAVOR, SUM(TOTAL_ORDER) AS TOTAL_ORDER
    FROM JULY
    GROUP BY FLAVOR
    ) AS J ON F.FLAVOR = J.FLAVOR
ORDER BY F.TOTAL_ORDER + J.TOTAL_ORDER DESC
LIMIT 3
```

```python
SELECT F.FLAVOR
FROM FIRST_HALF F
JOIN JULY J ON F.FLAVOR = J.FLAVOR
GROUP BY F.FLAVOR
ORDER BY (F.TOTAL_ORDER + SUM(J.TOTAL_ORDER)) DESC
LIMIT 3
```

### 2. 특정 기간동안 대여 가능한 자동차들의 대여비용 구하기

```python
SELECT CAR.CAR_ID, P.CAR_TYPE, FLOOR(CAR.DAILY_FEE * 30 * (100 - P.DISCOUNT_RATE) / 100) AS FEE
FROM CAR_RENTAL_COMPANY_DISCOUNT_PLAN AS P
JOIN (
    SELECT C.CAR_ID, C.CAR_TYPE, C.DAILY_FEE -- 종류랑 대여기간 만족
    FROM CAR_RENTAL_COMPANY_CAR AS C
    WHERE C.CAR_TYPE IN ('세단', 'SUV') -- 세단, SUV인 자동차
        AND NOT EXISTS (
            SELECT 1
            FROM CAR_RENTAL_COMPANY_RENTAL_HISTORY AS H
            WHERE H.CAR_ID = C.CAR_ID
                AND START_DATE <= '2022-11-30' AND END_DATE >= '2022-11-01'
        )    
    GROUP BY C.CAR_ID
    ) AS CAR ON CAR.CAR_TYPE = P.CAR_TYPE
WHERE P.DURATION_TYPE IN ('30일 이상')
    AND 500000 <= FLOOR(CAR.DAILY_FEE * 30 * (100 - P.DISCOUNT_RATE) / 100) AND FLOOR(CAR.DAILY_FEE * 30 * (100 - P.DISCOUNT_RATE) / 100) < 2000000
ORDER BY FEE DESC, CAR_TYPE ASC, CAR_ID DESC
```
